```python
import numpy as np
import math
import random

rk1beta = np.zeros((5,8))
rk2nssq = np.zeros((5,8,8))
rk2nssa = np.zeros((5,8))
rk2ss  =np.zeros((5,8,9))

#This function rounds down a bound for ch_2() to the nearest half-integer.
def halfint(x):
    return math.floor(2*x) / 2


print("When imbeta is rank 1:")
#When imbeta is rank 1, ch_2(Q^{-1}) <= 0 under assumptions.
for i in range(5):
    Ksq = 2*i+1
    for j in range(1,Ksq):
        if j > i-1:
            rk1beta[i][j-1] = -1
            #This case only exists when Kc_1(A) < K^2/2. A value of '-1' means this case is not possible.
        else:
            ch2imbeta = ((Ksq-1)**2)/(8*(Ksq))
            rk1beta[i][j-1] = halfint(ch2imbeta)
print(rk1beta)
#The rows correspond to K^2 from 1 to 9 (odd values only), and the columns correspond to Kc_1(A) values from 1 to 8.
            
print("")
print("")


print("When imbeta has tank 2 and is not slope-semistable:")

for i in range(5):
    Ksq = 2*i+1
    print(f"K^2 = {Ksq}")
    for j in range(2, Ksq-1):
         for k in range(j, Ksq-1):
            ch2imbeta = ((Ksq-1)**2)/(8*Ksq) + ((Ksq-3)**2)/(8*Ksq)
            ch2Q = (j-k)**2 / (2*Ksq)
            
            if k == j or ch2Q < 0.5:
            #If k = j, then imbeta is isomorphic to A.
                rk2nssq[i][j-2][k-2] = halfint(ch2imbeta)
            else:
                rk2nssq[i][j-2][k-2] = halfint(ch2imbeta) + halfint(ch2Q)
         print(f"Kc_1(A) = {j}")
         print(rk2nssq[i][j-2])
    print("")

print("")

print("When imbeta is rank 2 and slope-semistable:")
print("For each K^2 and Kc_1(A), rows represent bounds for every possible value of Kc_1(imbeta)")
for i in range(5):
    Ksq = 2*i+1
    print(f"K^2 = {Ksq}")
    for j in range(1,Ksq):
        print(f"Kc_1(A) = {j}")
        for k in range(j, Ksq+1):
            ch2imbeta = k**2 / (4*Ksq)
            ch2Q = (j-k)**2 / (2*Ksq)
            rk2ss[i][j-1][k-1] = halfint(ch2imbeta) + halfint(ch2Q)
#            print(f"({ch2imbeta}, {ch2Q})")


        print(rk2ss[i][j-1])
    print("")
```
